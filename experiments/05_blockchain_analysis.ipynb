# 05. Blockchain Metadata Storage & Integrity Verification
This notebook demonstrates Proof-of-Work block mining, metadata-only transaction records, and tamper verification.

```python
from src.pipeline import BiometricBlockchainPipeline
from src.biometric.landmark_extraction import LandmarkExtractor

pipeline = BiometricBlockchainPipeline()
extractor = LandmarkExtractor()
landmarks = extractor.generate_synthetic_subject_landmarks(subject_id=1)

# Process and store on blockchain
res = pipeline.process_end_to_end(landmarks, "Sensitive Financial Record", tx_id="tx_999")
print(f"Mined Block #{res['block_index']} with hash {res['block_hash']}")
print(f"Verification status: {res['verification_msg']}")

# Demonstrate Tamper Detection
tampered_ciphertext = bytearray(res['encrypted_result']['combined'])
tampered_ciphertext[0] ^= 0x01
is_valid, msg, _ = pipeline.verifier.verify_ciphertext_integrity("tx_999", bytes(tampered_ciphertext))
print(f"Tampered Ciphertext Verification Result: {is_valid} -> {msg}")
```
