# 04. AES-256-GCM Encryption and Tamper Detection Benchmarking
This notebook demonstrates authenticated encryption, decryption, and tamper detection using biometric-derived keys.

```python
from src.encryption.aes import AESGCMCipher
import os

key = os.urandom(32)
cipher = AESGCMCipher(key)

plaintext = "Sensors and Blockchain Confidential State Data Payload"
encrypted = cipher.encrypt(plaintext)
print(f"Nonce: {encrypted['nonce'].hex()}")
print(f"Ciphertext: {encrypted['ciphertext'].hex()}")
print(f"Auth Tag: {encrypted['tag'].hex()}")

# Decrypt Valid Payload
decrypted = cipher.decrypt(encrypted).decode('utf-8')
print(f"Decrypted String: '{decrypted}' (Success: {decrypted == plaintext})")

# Tamper Demonstration
tampered_combined = bytearray(encrypted['combined'])
tampered_combined[5] ^= 0xFF
try:
    cipher.decrypt({'nonce': encrypted['nonce'], 'combined': bytes(tampered_combined)})
except ValueError as e:
    print(f"Tamper Caught Successfully: {e}")
```
