# 02. Genetic Algorithm (GA) Feature Selection Analysis
This notebook analyzes the convergence, fitness progression, and feature selection mask of the Genetic Algorithm.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.biometric.landmark_extraction import LandmarkExtractor
from src.biometric.preprocessing import FeatureNormalizer
from src.optimization.genetic_algorithm import GeneticAlgorithmOptimizer

# Generate 5 subject captures
extractor = LandmarkExtractor()
normalizer = FeatureNormalizer()

captures_dict = {}
for sub in range(1, 6):
    sub_caps = []
    for cap in range(3):
        lm = extractor.generate_synthetic_subject_landmarks(sub, cap, noise_level=0.02)
        sub_caps.append(normalizer.normalize_landmarks(lm))
    captures_dict[sub] = sub_caps

# Run GA Optimization
ga = GeneticAlgorithmOptimizer()
best_mask, history = ga.optimize(feature_dim=212, subject_captures=captures_dict)

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(history) + 1), history, marker='o', color='purple')
plt.title("Genetic Algorithm Fitness Progression")
plt.xlabel("Generation")
plt.ylabel("Best Multi-Objective Fitness")
plt.grid(True)
plt.show()

print(f"Selected Features Count: {np.sum(best_mask)} / {len(best_mask)}")
```
