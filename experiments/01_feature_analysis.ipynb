# 01. Facial Landmark Feature Analysis
This notebook analyzes the 106 facial landmark extraction and Procrustes feature normalization pipeline.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.biometric.landmark_extraction import LandmarkExtractor
from src.biometric.preprocessing import FeatureNormalizer

# 1. Extract 106 Landmark Coordinates
extractor = LandmarkExtractor(num_landmarks=106)
landmarks = extractor.generate_synthetic_subject_landmarks(subject_id=1)
print(f"Extracted Landmarks Shape: {landmarks.shape}")

# 2. Plot Raw vs Normalized Landmarks
normalizer = FeatureNormalizer()
norm_features = normalizer.normalize_landmarks(landmarks)
norm_pts = norm_features.reshape(-1, 2)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(landmarks[:, 0], landmarks[:, 1], c='blue', s=15)
plt.gca().invert_yaxis()
plt.title("Raw Facial Landmarks (106 points)")

plt.subplot(1, 2, 2)
plt.scatter(norm_pts[:, 0], norm_pts[:, 1], c='red', s=15)
plt.title("Procrustes Normalized Landmarks")
plt.tight_layout()
plt.show()
```
